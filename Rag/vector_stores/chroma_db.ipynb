```python
from langchain_classic.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
model_name = 'sentence-transformers/all-MiniLM-L6-V2'
embedding_model = HuggingFaceEmbeddings(model_name = model_name)
```

```python
print(embedding_model)
```

```python
vector_store = Chroma(
    embedding_function = embedding_model,
    persist_directory = r"D:\live_Projects\langchain\Rag\vector_stores\chroma_db",
    collection_name = 'sample_chroma_db'
)
```

```python
vector_store.add_documents(documents = documents, 
                           ids = [str(i + 1) for i in range(len(documents))])
```

```python
# fetch the details from store

vector_store.get(include = ['embeddings'])
```

```python
query = input("Enter your question : ")
query_embedding = embedding_model.embed_query(query)
print(query_embedding)
```

```python
def fetch_doc_id(search_text):
    data = vector_store.get()
    target_id = None

    for i , doc in enumerate(data['documents']):
        if str(search_text).lower() in doc.lower():
            target_id = data['ids'][i]
            break
    return target_id
```

```python
vector_store.get()
```

```python
target_id = fetch_doc_id(document_1.page_content)
```

```python
vector_store.update_document(document= new_doc, document_id = target_id)
```
